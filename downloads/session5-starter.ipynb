```python
import torch
import matplotlib.pyplot as plt
drop = [0.50, 0.75, 1.00, 1.25, 1.50, 1.75, 2.00]
bounce = [0.31, 0.45, 0.60, 0.73, 0.89, 1.02, 1.18]
x = torch.tensor(drop)
y = torch.tensor(bounce)
```

## 1. One Step by Hand
gradient for w = mean of 2 x error x drop, gradient for b = mean of 2 x error

```python
w, b = 0.0, 0.0
learning_rate = 0.1
error = (w * x + b) - y
grad_w = (2 * error * x).mean().item()
grad_b = (2 * error).mean().item()
print("gradient for w:", round(grad_w, 4), "  gradient for b:", round(grad_b, 4))
w = w - learning_rate * grad_w
b = b - learning_rate * grad_b
print("after one step: w", round(w, 4), " b", round(b, 4))
```

## 2. Let PyTorch Find the Gradients

```python
w = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
loss = ((w * x + b - y) ** 2).mean()
loss.backward()
print("PyTorch gradient for w:", round(w.grad.item(), 4), "  for b:", round(b.grad.item(), 4))
```

## 3. Train It

```python
def fit(learning_rate, steps=200, show_every=40):
    w = torch.tensor(0.0, requires_grad=True)
    b = torch.tensor(0.0, requires_grad=True)
    for step in range(steps + 1):
        loss = ((w * x + b - y) ** 2).mean()     # 1. how wrong are we?
        if step % show_every == 0:
            print("step", step, " w", round(w.item(), 4), " b", round(b.item(), 4), " loss", round(loss.item(), 6))
        loss.backward()                          # 2. find the gradients
        with torch.no_grad():
            w -= learning_rate * w.grad          # 3. step downhill
            b -= learning_rate * b.grad
        w.grad.zero_()                           # 4. wipe the gradients
        b.grad.zero_()
    return w.item(), b.item()

w_learned, b_learned = fit(learning_rate=0.1, steps=1000, show_every=200)
```

## 4. Check Against the Exact Answer

```python
w_exact = ((x - x.mean()) * (y - y.mean())).sum() / ((x - x.mean()) ** 2).sum()
b_exact = y.mean() - w_exact * x.mean()
print("gradient descent: w", round(w_learned, 4), " b", round(b_learned, 4))
print("least squares:    w", round(w_exact.item(), 4), " b", round(b_exact.item(), 4))
```

## 6. Why Wipe the Gradients?

```python
w = torch.tensor(0.0, requires_grad=True)
for i in range(3):
    loss = ((w * x - y) ** 2).mean()
    loss.backward()
    print("stored gradient:", round(w.grad.item(), 4))   # it keeps growing: gradients add up
```

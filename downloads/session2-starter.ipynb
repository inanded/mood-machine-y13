## 2. Painting a Tiny Picture
Square brackets pick out parts of a tensor, counting from 0. A colon `:` means "all of them".
The first number picks the colour grid: 0 = red, 1 = green, 2 = blue.

```python
import matplotlib.pyplot as plt

picture = torch.zeros(3, 4, 4)   # start all black
picture[0] = 1.0                 # turn the whole red grid up full
picture[1, 0, :] = 1.0           # green grid, row 0, every column: red + green = yellow
print(picture[:, 0, 0])          # the top-left pixel
print(picture[:, 3, 0])          # the bottom-left pixel

plt.imshow(picture.permute(1, 2, 0))   # permute puts colour last, which is how pictures are drawn
plt.axis("off")
plt.show()
```

## 5. Finding the Biggest Score
The Mood Machine will give 10 scores. Its guess is the position of the biggest one.

```python
scores = torch.tensor([0.2, -1.3, 0.1, 3.4, 0.0, 0.7, 1.1, -0.2, 0.5, 0.9])
best = scores.argmax().item()
print("biggest score is at position", best, "which is", CLASSES[best])
```

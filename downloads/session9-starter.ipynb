```python
import torch
x = torch.tensor([1.0, 2.0])
W = torch.tensor([[0.5, 0.25], [-0.5, 0.5]], requires_grad=True)   # inputs -> hidden
v = torch.tensor([1.0, 2.0], requires_grad=True)                     # hidden -> output
target = torch.tensor(3.0)
```

## 1. Forward Pass

```python
z = W @ x                  # each hidden neuron's total
h = torch.relu(z)          # after ReLU
out = v @ h                # the output
loss = (out - target) ** 2
print("z:", z.tolist())
print("h:", h.tolist())
print("out:", out.item(), "  loss:", loss.item())
```

## 3. Update the Weights

```python
def loss_after_step(learning_rate):
    with torch.no_grad():
        W2 = W - learning_rate * W.grad
        v2 = v - learning_rate * v.grad
        new_out = v2 @ torch.relu(W2 @ x)
        return ((new_out - target) ** 2).item()

print("loss before:", loss.item())
print("loss after one step with learning rate 0.01:", round(loss_after_step(0.01), 4))
```

## 4. The Same Thing Inside MoodNet
`loss.backward()` in Session 8 did exactly this, for 820,874 weights at once.

```python
from torch import nn
torch.manual_seed(0)
small = nn.Sequential(nn.Linear(2, 2), nn.ReLU(), nn.Linear(2, 1))
guess = small(x)
loss2 = (guess - target) ** 2
loss2.backward()
for name, p in small.named_parameters():
    print(name, "gradient:", [round(g, 3) for g in p.grad.flatten().tolist()])
```

```python
import torch
from torch import nn
import matplotlib.pyplot as plt
```

## 1. One Neuron
Multiply each input by its weight, add them up, add the bias, then apply ReLU.

```python
x = torch.tensor([0.5, 0.8])      # inputs
w = torch.tensor([0.4, -0.6])     # weights
b = 0.1                           # bias

weighted = w * x
print("each input times its weight:", [round(v, 2) for v in weighted.tolist()])
total = weighted.sum() + b
print("total with the bias:", round(total.item(), 2))
print("after ReLU:", torch.relu(total).item())
```

## 2. What the Bias Does
The bias is added after the weighted inputs. It shifts how easily the neuron switches on.

```python
for bias in [-0.5, 0.1, 0.5]:
    total = (w * x).sum() + bias
    print("bias", bias, " total", round(total.item(), 2), " output", round(torch.relu(total).item(), 2))
```

## 3. ReLU
ReLU keeps positive totals and turns negative ones into 0.

```python
totals = torch.linspace(-3, 3, 61)
plt.plot(totals, torch.relu(totals))
plt.xlabel("total going in")
plt.ylabel("ReLU output")
plt.grid(alpha=0.3)
plt.show()
print(torch.relu(torch.tensor([-2.0, -0.18, 0.0, 0.5, 3.0])))
```

## 4. Why ReLU Matters
Two layers with no activation between them can always be squashed into one layer. ReLU stops that.

```python
torch.manual_seed(0)
layer1 = nn.Linear(2, 3)
layer2 = nn.Linear(3, 1)
x = torch.tensor([0.5, 0.8])

# without ReLU: layer2(layer1(x)) is just one bigger straight-line sum
W = layer2.weight @ layer1.weight
B = layer2.weight @ layer1.bias + layer2.bias
print("two layers, no ReLU:", round(layer2(layer1(x)).item(), 4))
print("one combined layer: ", round((W @ x + B).item(), 4))

# with ReLU in between, no single layer can do the same for every input
print("two layers with ReLU:", round(layer2(torch.relu(layer1(x))).item(), 4))
```

## 5. A Layer of Neurons
Each row of `W` is one neuron's weights. `W @ x` works out every neuron at once.

```python
W = torch.tensor([[0.4, -0.6],
                  [0.9,  0.2],
                  [-0.3, 0.5]])         # 3 neurons, 2 inputs each
b = torch.tensor([0.1, 0.0, -0.2])
print("3 neurons give:", torch.relu(W @ x + b))
```

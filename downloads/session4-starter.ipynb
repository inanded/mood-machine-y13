```python
# YOUR TURN: replace these with your own measurements, in metres.
# The backup values are from a tennis ball dropped onto concrete.
drop = [0.50, 0.75, 1.00, 1.25, 1.50, 1.75, 2.00]
bounce = [0.31, 0.45, 0.60, 0.73, 0.89, 1.02, 1.18]
```

```python
import torch
import matplotlib.pyplot as plt
x = torch.tensor(drop)
y = torch.tensor(bounce)
plt.scatter(x, y)
plt.xlabel("drop height (m)")
plt.ylabel("bounce height (m)")
plt.show()
```

## 1. A Model: A Line
`predicted bounce = w * drop + b`

```python
def predict(w, b):
    return w * x + b

w, b = 0.5, 0.0
print("predictions:", [round(p, 2) for p in predict(w, b).tolist()])
print("measured:   ", bounce)
```

## 2. The Loss: Mean Squared Error

```python
def mse(w, b):
    errors = predict(w, b) - y          # 1. the errors
    squared = errors ** 2               # 2. square each one
    return squared.mean()               # 3. take the mean

errors = predict(0.5, 0.0) - y
print("errors:", [round(e, 3) for e in errors.tolist()])
print("loss for w=0.5, b=0:", round(mse(0.5, 0.0).item(), 5))
```
